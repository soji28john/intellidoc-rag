```python

from src.rag_pipeline import RAGPipeline
from src.evaluation import evaluate_retrieval, evaluate_generation
rag = RAGPipeline()
print("RAG Pipeline initialized.")
test_questions = [
    "What is the objective of the project?",
    "Who is the target audience?",
    "What methodology is being used?"
]

ground_truth_keywords = [
    ["objective", "goal", "purpose"],
    ["audience", "users"],
    ["methodology", "approach", "process"]
]

reference_answers = [
    "The project objective is to create an intelligent document assistant.",
    "The target audience is enterprise knowledge workers.",
    "The methodology uses RAG and LLM techniques."
]

print(evaluate_retrieval(rag, test_questions, ground_truth_keywords))
print(evaluate_generation([rag.query(q)["answer"] for q in test_questions], reference_answers))

```

```python
# Generate answers via RAG
generated_answers = [rag.query(q) for q in test_questions]

print("Retrieval Evaluation:")
print(evaluate_retrieval(rag, test_questions, ground_truth_keywords))

print("\nGeneration Evaluation:")
print(evaluate_generation(generated_answers, reference_answers))
```
