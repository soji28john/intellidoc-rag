```python
import sys, os
sys.path.append(os.path.abspath(".."))

from src.document_processor import DocumentProcessor
from src.embeddings import EmbeddingGenerator
from src.vector_store import VectorStore

# Step 1: Process documents
processor = DocumentProcessor()
docs = processor.process_document('../data/raw/sample.pdf')

# Step 2: Generate embeddings
embedding_gen = EmbeddingGenerator()
texts = [doc['content'] for doc in docs]
embeddings = embedding_gen.generate_embeddings_batch(texts)

# Step 3: Store in vector database
vector_store = VectorStore()
vector_store.add_documents(
    documents=texts,
    embeddings=embeddings.tolist(),
    metadatas=[doc['metadata'] for doc in docs]
)

print(f"✓ Stored {len(docs)} document chunks")
print(f"✓ Collection stats: {vector_store.get_collection_stats()}")

# Step 4: Test retrieval
query = "What is the main topic of this document?"
query_embedding = embedding_gen.generate_embedding(query)
results = vector_store.search(query_embedding, n_results=3)

print(f"\nQuery: {query}")
print(f"\nTop 3 relevant chunks:")
for i, doc in enumerate(results['documents'][0]):
    print(f"\n--- Result {i+1} ---")
    print(doc[:200] + "...")
```
