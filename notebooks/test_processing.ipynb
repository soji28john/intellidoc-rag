```python
# Test the document processor
import sys, os
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), "..")))

from src.document_processor import DocumentProcessor
print(os.getcwd())


# Initialize processor
processor = DocumentProcessor(chunk_size=500, chunk_overlap=100)

# Test with a sample PDF document
docs = processor.process_document('../data/raw/sample.pdf')

# Analyze results
print(f"Total chunks created: {len(docs)}")
print(f"\nFirst chunk preview:\n{docs[0]['content'][:200]}...")
print(f"\nMetadata: {docs[0]['metadata']}")

```

```python
# Visualize chunk sizes
import matplotlib.pyplot as plt

chunk_lengths = [len(doc['content']) for doc in docs]
plt.hist(chunk_lengths, bins=20)
plt.xlabel('Chunk Length (characters)')
plt.ylabel('Frequency')
plt.title('Distribution of Chunk Sizes')
plt.show()
```
